### 1.1.2 VLT

```julia
vline!([50], label="VLT Line at x=50", color=:red, linestyle=:dash, lw=2)
```

```julia

x_var = 0:0.1:100
y_p = sqrt.(x_var)
y_n = -sqrt.(x_var)

plot(x_var, y_p, label="y = +√x" ,color=:gray)
plot!(x_var, y_n, label="y = -√x" ,color=:gray)

vline!([25], label="VLT Line at x=25", color=:red, linestyle=:dash, lw=2)
```

### 1.1.3 Example Function

#### f = x^2 ( LATEX )

```julia
x = -10:0.1:10
f(x) = x.^2

plot(x, f(x), 
     label="f(x) = x^2", 
     title="Symmetric Quadratic Curve around (0,0)", 
     color=:purple, 
     lw=2,
     framestyle=:origin,
     grid=true)
```

#### f = x (LATEX)

```julia
x = -100:0.1:100
f(x) = x

plot(x, f(x), 
     label="f(x) = x",
     color=:purple, 
     lw=2,
     framestyle=:origin,
     grid=true)
```

#### f = x^3 (LATEX)

```julia
x = -10:0.1:10
f(x) = x.^3

plot(x, f(x), 
     label="f(x) = x.^3",
     color=:purple, 
     lw=2,
     framestyle=:origin,
     grid=true)
```

#### f = sqrt(x) (LATEX)

```julia
x = 0:0.1:10
z = sqrt.(x)

plot(x, z, 
     label="f(x) = sqrt(x)",
     color=:purple, 
     lw=2,
     framestyle=:origin,
     grid=true)
```
