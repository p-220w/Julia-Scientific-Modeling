## 1.1 Function Basics

<a id="define-a-function"></a>
### 1.1.1 Define a Function

**Mathematical Concept:**
A function assigns exactly one output to each input domain value.
$$ f(x) = x^2 $$

**Modeling Objective:**
Programmatically evaluate and visualize a basic mathematical mapping.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
f(x) = x.^2
```

```julia
xV = 1:100
plot(xV,f(xV), title = "WOW")
```

<a id="vertical-line-test"></a>
### 1.1.2 Vertical Line Test (VLT)

**Mathematical Concept:**
A relation is a function if any vertical line intersects its graph at most once.
$$ y^2 = x \implies y = \pm\sqrt{x} $$

**Modeling Objective:**
Geometrically demonstrate a non-function relation using vertical line intersections.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
vline!([50], label="VLT Line at x=50", color=:red, linestyle=:dash, lw=2)
```

```julia

x_var = 0:0.1:100
y_p = sqrt.(x_var)
y_n = -sqrt.(x_var)

plot(x_var, y_p, label="y = +√x" ,color=:gray)
plot!(x_var, y_n, label="y = -√x" ,color=:gray)

vline!([25], label="VLT Line at x=25", color=:red, linestyle=:dash, lw=2)
```

<a id="example-functions"></a>
### 1.1.3 Example Functions

<a id="example-quadratic"></a>
#### Quadratic Function

**Mathematical Concept:**
A symmetric, second-degree polynomial curve forming a parabola.
$$ f(x) = x^2 $$

**Modeling Objective:**
Plot symmetric non-linear growth around the origin.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = x.^2

plot(x, f(x), 
     label="f(x) = x^2", 
     title="Symmetric Quadratic Curve around (0,0)", 
     color=:purple, 
     lw=2,
     framestyle=:zerolines,
     grid=true)
```

<a id="example-linear"></a>
#### Linear Function

**Mathematical Concept:**
A first-degree polynomial representing a constant rate of change.
$$ f(x) = x $$

**Modeling Objective:**
Visualize direct proportionality and linear mapping.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
x = -100:0.1:100
f(x) = x

plot(x, f(x), 
     label="f(x) = x",
     color=:purple, 
     lw=2,
     framestyle=:zerolines,
     grid=true)
```

<a id="example-cubic"></a>
#### Cubic Function

**Mathematical Concept:**
A third-degree odd polynomial with origin symmetry.
$$ f(x) = x^3 $$

**Modeling Objective:**
Illustrate unbounded accelerated growth and point symmetry.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = x.^3
plotly()
plot(x, f(x), 
     label="f(x) = x.^3",
     color=:purple, 
     lw=2,
     framestyle=:zerolines,
     grid=true)
```

<a id="example-sqrt"></a>
#### Square Root Function

**Mathematical Concept:**
The principal inverse of $x^2$, restricted to non-negative reals.
$$ f(x) = \sqrt{x} $$

**Modeling Objective:**
Showcase domain restriction and decelerating positive growth.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
x = 0:0.1:10
z = sqrt.(x)

plot(x, z, 
     label="f(x) = sqrt(x)",
     color=:purple, 
     lw=2,
     framestyle=:zerolines,
     grid=true)
```

#### More Examples

<a id="multivariable-paraboloid"></a>
#### 3D Paraboloid

**Mathematical Concept:**
A multivariable quadratic function forming a symmetric 3D surface.
$$ f(x, y) = x^2 + y^2 $$

**Modeling Objective:**
Render a 3D surface plot to visualize continuous spatial domains.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
f(x,y) = x^2 + y^2
x = -5:0.2:5
y = -5:0.2:5

surface(x,y,f, title= "3D Visul of Multivariable", color=:coolwarm, camera=(40,35), zlims=(0, 25) )
```

```julia
f(x,y) = x^2 + y^2
x = -5:0.2:5
y = -5:0.2:5

surface(x,y,f, title= "3D Visul of Multivariable WITHOUT zlims", color=:coolwarm, camera=(40,35))
```
