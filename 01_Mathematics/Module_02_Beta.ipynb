## 2.1 Equations of Lines and Quadratics

### 2.1.1 Equations of Lines

#### 2.1.1.1 Slope

#### y = mx + b (LATEX)

```julia
p1 = (x1, y1) = (2.0, 5.0)
p2 = (x2, y2) = (6.0, 13.0)

m = (y2 - y1) / (x2 - x1)
b = y1 - m * x1
line_model(x) = m * x + b
x_range = -2:0.1:10

plot(x_range, line_model.(x_range), 
     title="Core Model: Linear Equation y = mx + b",
     label="m",
     color=:purple,
     framestyle=:zerolines,)

scatter!([x1, x2], [y1, y2], 
         color=:red,
         label="Points")

scatter!([0],[b],
         color=:blue,
         label="b")
```

### 2.1.2 Quadratic Functions

#### f(x) = a*x^2 + b*x + c (LATEX)

#### 2.1.2.1 Parabolas

```julia
@syms x
# let c > 0
c = 3
f = (x + c)^2

#fix plot code
hh = -c
margin = 2c + 1 
x_left  = hh - margin
x_right = hh + margin
x_domain = x_left:0.1:x_right

plot(x_domain,f, title="shift to left" , color=:purple, framestyle=:zerolines,
 xlims = (x_left, x_right))
```

```julia
@syms x
# let c > 0
c = 3
f = (x - c)^2

#fix plot code
hh = c
margin = 2c + 1 
x_left  = hh - margin
x_right = hh + margin
x_domain = x_left:0.1:x_right

plot(x_domain,f, title="shift to right" , color=:purple, framestyle=:zerolines,
 xlims = (x_left, x_right))
```

```julia
@syms x
# let c > 0
c = 3
f = (x + c)^2 + 5

#fix plot code
hh = -c
margin = 2c + 1 
x_left  = hh - margin
x_right = hh + margin
x_domain = x_left:0.1:x_right

plot(x_domain,f, title="shift to left and up" , color=:purple, framestyle=:zerolines,
 xlims = (x_left, x_right))
```

```julia
@syms x
# let c > 0
c = 3
f = (x - c)^2 - 5

#fix plot code
hh = c
margin = 2c + 1 
x_left  = hh - margin
x_right = hh + margin
x_domain = x_left:0.1:x_right

plot(x_domain,f, title="shift to right and dowm" , color=:purple, framestyle=:zerolines,
 xlims = (x_left, x_right))
```

#### 2.1.2.2 Vertex

#### x = -b/2a (LATEX)
#### y = (4ac-b^2) / 4a ( LATEX )

```julia
@syms x

f = x^2 + x + 1
a = f.coeff(x,2)
b = f.coeff(x,1)
c = f.coeff(x,0)
vx = -b/2a
vy = f(x => vx)

#fix plot code
margin = 2 * abs(vx) + 2
x_left  = vx - margin
x_right = vx + margin
x_domain = x_left:0.1:x_right

plot(x_domain,f, title="Vertax" , color=:purple, framestyle=:zerolines,
 xlims = (x_left, x_right))

 scatter!([vx],[vy] , color=:gold, ms=6, label=" Vartex ($vx , $vy )")
```
