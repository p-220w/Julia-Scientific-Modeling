## 3.1 Exponential Functions

### 3.1.1

### f = base^x

### Different between , f = x^2 , f = a^x

(LATEX)

```julia
x = -10:0.1:10
f(x) = x^2

plot(x,f,color=:purple, framestyle=:zerolines, title="x^2")
```

```julia
x = -10:0.1:10
f(x) = 2^x

plot(x,f,color=:purple, framestyle=:zerolines, title="2^x")
```

### 3.1.2 Kinds of Exponential Functions

#### f = (1/a)^x , 0<a<1 (LATEX)

```julia
x = -10:0.1:10
f(x) = (1/2)^x

plot(x,f,color=:purple, framestyle=:zerolines, title="(1/a)^x")
```

#### f = 1^x , a = 1 (LATEX) 

```julia
x = -10:0.1:10
f(x) = 1^x

plot(x,f,color=:purple, framestyle=:zerolines, title="1^x")
```

#### f = a^x (LATEX) a>1 (LATEX)

```julia
x = -10:0.1:10
f(x) = 2^x

plot(x,f,color=:purple, framestyle=:zerolines, title="a^x")
```

#### Some Example

```julia
x = -10:0.1:10
f(x) = -2^x-2 - 50

plot(x,f,color=:purple, framestyle=:zerolines, title="-2^x-2 - 50")
```

### 3.1.4 Examples of "e" 

#### f = (1-e^x)/(1-e^(1-e^2)) (LATEX WITHOUT)

```julia
@syms x e
f_e = (1-e^x)/(1-e^(1-x^2))
f_demon = (1-e^(1-x^2))
sol = solve(f_demon, x)
display(sol)
display("D: R/{-1,+1}")
```

#### f = 5^x

#### find ( f(x+h) - f(x) / h )

(LATEX WITHOUT)

```julia
@syms x,h

fx(x) = 5^x
f_sol = ( fx(x+h) - fx(x) / h )
factor(f_sol)
```

#### f = (1-e^(1/x))/(1=e^(1/x)) (LATEX WITHOUT)

```julia
@syms x,e

f_ex = (1-e^(1/x))/(1+e^(1/x))
factor(f_ex)
```
