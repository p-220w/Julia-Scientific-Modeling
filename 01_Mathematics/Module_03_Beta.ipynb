### 3.2.2 Logarithmic Rules

Definition$\log_a(x) = y \iff a^y = x$Product Rule$\log_a(xy) = \log_a(x) + \log_a(y)$Quotient Rule$\log_a\left(\frac{x}{y}\right) = \log_a(x) - \log_a(y)$Power Rule$\log_a(x^r) = r \cdot \log_a(x)$Identity & Zero Rules$\log_a(a) = 1$$\log_a(1) = 0$$\log_a(0) = \text{undefined}$Inverse Properties$\log_a(a^x) = x$$a^{\log_a(x)} = x$Change of Base Formula$\log_a(x) = \frac{\ln(x)}{\ln(a)} = \frac{\log_{10}(x)}{\log_{10}(a)}$

(LATEX)

#### Examples

#### log(5,125) = x , find x (LATEX WITHOUT)

```julia
@syms x
eq_log = log(5,125) ~ x
sol = solve(eq_log,x)
```

```julia
@syms a,b

eq_logs = log(a,b) - log(a,b) + log(a,b)
display(simplify(eq_logs))

eq_logs = log(2,6) - log(2,15) + log(2,20)
display(eq_logs)
```

### 3.3.2 Natural Logarithm Rules

Definition$\ln(x) = y \iff e^y = x$Inverse Properties$\ln(e^x) = x$$e^{\ln(x)} = x$Identity & Exponential Base Rules$\ln(e) = 1$$e^0 = 1$Zero & Undefined Rules$\ln(1) = 0$$\ln(0) = \text{undefined}$Product Rule$\ln(xy) = \ln(x) + \ln(y)$Quotient Rule$\ln\left(\frac{x}{y}\right) = \ln(x) - \ln(y)$Power Rule$\ln(x^r) = r \cdot \ln(x)$

(LATEX)
