## 3.1 Exponential Functions

<a id="polynomial-vs-exponential"></a>
### 3.1.1 Polynomial vs Exponential Functions

**Mathematical Concept:**
Polynomial functions feature a variable base with a constant exponent, whereas exponential functions maintain a constant base with a variable exponent.
$$ f(x) = x^2 \quad \text{vs} \quad f(x) = a^x $$

**Modeling Objective:**
Contrast the differing rates of change and graphical trajectories between polynomial and exponential growth models.

[↑ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = x^2

plot(x,f,color=:purple, framestyle=:zerolines, title="x^2")
```

```julia
x = -10:0.1:10
f(x) = 2^x

plot(x,f,color=:purple, framestyle=:zerolines, title="2^x")
```

<a id="kinds-of-exponential-functions"></a>
### 3.1.2 Kinds of Exponential Functions

<a id="exponential-decay"></a>
#### Exponential Decay

**Mathematical Concept:**
When the base $a$ is strictly between 0 and 1, the function models exponential decay, asymptotically approaching the x-axis as $x$ increases.
$$ f(x) = \left(\frac{1}{a}\right)^x , \quad 0 < \frac{1}{a} < 1 $$

**Modeling Objective:**
Visualize continuous decay within a system over a progressive domain.

[↑ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = (1/2)^x

plot(x,f,color=:purple, framestyle=:zerolines, title="(1/a)^x")
```

<a id="constant-exponential"></a>
#### Constant Exponential Function

**Mathematical Concept:**
An exponential function with a base of exactly 1 degenerates into a constant horizontal line, as 1 raised to any power remains 1.
$$ f(x) = 1^x = 1 $$

**Modeling Objective:**
Establish a neutral mathematical baseline to visually separate exponential growth regimes from decay regimes.

[↑ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = 1^x

plot(x,f,color=:purple, framestyle=:zerolines, title="1^x")
```

<a id="exponential-growth"></a>
#### Exponential Growth

**Mathematical Concept:**
If the base $a$ is strictly greater than 1, the function dictates exponential growth, escalating at an accelerating rate for positive inputs.
$$ f(x) = a^x , \quad a > 1 $$

**Modeling Objective:**
Demonstrate rapid, unbounded acceleration characteristic of compound growth models.

[↑ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = 2^x

plot(x,f,color=:purple, framestyle=:zerolines, title="a^x")
```

<a id="exponential-transformations"></a>
#### Exponential Transformations

**Mathematical Concept:**
Exponential curves can undergo geometric transformations including reflections across axes and vertical/horizontal translations.
$$ f(x) = -2^{x-2} - 50 $$

**Modeling Objective:**
Model a compound mathematical transformation combining a downward reflection and a negative vertical shift.

[↑ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = -2^x-2 - 50

plot(x,f,color=:purple, framestyle=:zerolines, title="-2^x-2 - 50")
```

<a id="eulers-number"></a>
### 3.1.3 Euler's Number (e)

**Mathematical Concept:**
Euler's number, $e$, is an irrational mathematical constant fundamental to describing continuous growth and natural decay systems.
$$ e = \lim_{n \to \infty} \left(1 + \frac{1}{n}\right)^n \approx 2.71828 $$

[↑ Back to Table of Contents](#table-of-contents)

<a id="examples-of-e"></a>
### 3.1.4 Examples of "e"

<a id="example-e-1"></a>
#### Example 1: Rational Exponential Function
$$ f(x) = \frac{1 - e^x}{1 - e^{1-x^2}} $$

[↑ Back to Table of Contents](#table-of-contents)

```julia
@syms x e
f_e = (1-e^x)/(1-e^(1-x^2))
f_demon = (1-e^(1-x^2))
sol = solve(f_demon, x)
display(sol)
display("D: R/{-1,+1}")
```

<a id="example-e-2"></a>
#### Example 2: Difference Quotient
$$ \frac{f(x+h) - f(x)}{h} = \frac{5^{x+h} - 5^x}{h} $$

[↑ Back to Table of Contents](#table-of-contents)

```julia
@syms x,h

fx(x) = 5^x
f_sol = ( fx(x+h) - fx(x) / h )
factor(f_sol)
```

<a id="example-e-3"></a>
#### Example 3: Reciprocal Exponent
$$ f(x) = \frac{1 - e^{\frac{1}{x}}}{1 + e^{\frac{1}{x}}} $$

[↑ Back to Table of Contents](#table-of-contents)

```julia
@syms x,e

f_ex = (1-e^(1/x))/(1+e^(1/x))
factor(f_ex)
```

## 3.2 Logarithmic Functions

<a id="logarithmic-basics"></a>
### 3.2.1 Logarithmic Basics

**Mathematical Concept:**
Logarithms are the mathematical inverse of exponentiation, calculating the specific exponent required to reach a particular value from a given base.
$$ y = \log_a(x) \quad , \quad y = \ln(x) = \log_e(x) $$

[↑ Back to Table of Contents](#table-of-contents)

<a id="logarithm-exponential-inverse"></a>
#### Logarithm and Exponential Inverse Relationship

**Mathematical Concept:**
A logarithm identically reflects its corresponding exponential function across the line of symmetry $y = x$.
$$ \log_a(x) = y \iff a^y = x $$

**Modeling Objective:**
Visually validate the strict geometric mirror symmetry between $2^x$ and $\log_2(x)$.

[↑ Back to Table of Contents](#table-of-contents)

```julia
x_log = 0.1:0.1:10
f_log(x) = log(2,x)
plot(x_log,f_log,color=:purple,label="f_log", framestyle=:zerolines, xlims=(-3, 4), ylims=(-3, 4),aspect_ratio=:equal)

x_pw = -10:0.1:10
f_pw(x) = 2^x
plot!(x_pw,f_pw,color=:purple,label="f_pw", framestyle=:zerolines, lw=0.2)

plot!(-3:4, -3:4, label="(Mirror)", color=:red, linestyle=:dash, lw=1.5)
```

<a id="logarithm-properties"></a>
#### Logarithmic Function Properties

* **Domain:** $x \in (0, \infty)$
* **Range:** $y \in (-\infty, \infty)$
* **x-intercepts:** $(1, 0)$
* **y-intercepts:** None

[↑ Back to Table of Contents](#table-of-contents)

<a id="logarithmic-rules"></a>
### 3.2.2 Logarithmic Rules

<a id="logarithmic-algebraic-rules"></a>
#### Logarithmic Algebraic Axioms

**Mathematical Concept:**
A rigid set of algebraic axioms governing logarithms, allowing complex products and quotients to be simplified into sums and differences.
$$
\begin{aligned}
\log_a(xy) &= \log_a(x) + \log_a(y) \\
\log_a\left(\frac{x}{y}\right) &= \log_a(x) - \log_a(y) \\
\log_a(x^r) &= r \cdot \log_a(x) \\
\log_a(a) &= 1 \quad , \quad \log_a(1) = 0 \\
a^{\log_a(x)} &= x \\
\log_a(x) &= \frac{\ln(x)}{\ln(a)} = \frac{\log_{10}(x)}{\log_{10}(a)}
\end{aligned}
$$

## 3.3 Natural Logarithm

<a id="natural-logarithm-ln"></a>
### 3.3.1 Natural Logarithm (ln)

<a id="ln-definition"></a>
#### Definition
$$ \ln(x) = \log_e(x) $$

[↑ Back to Table of Contents](#table-of-contents)

<a id="ln-inverse-relationship"></a>
#### The Natural Logarithm Inverse

**Mathematical Concept:**
The natural logarithm ($\ln$) specifically unwinds continuous exponential growth ($e^x$), mirroring perfectly across the line $y = x$.
$$ \ln(x) = y \iff e^y = x $$

**Modeling Objective:**
Render continuous exponential and natural logarithmic graphs to highlight their inherent base-$e$ symmetry.

[↑ Back to Table of Contents](#table-of-contents)

```julia
x_exp = -3:0.05:3
f_exp(x) = exp.(x)
plot(x_exp,f_exp,color=:purple,label="f_exp", framestyle=:zerolines, xlims=(-3, 4), ylims=(-3, 4),aspect_ratio=:equal, lw=0.25)

x_log = 0.05:0.05:3
f_log(x) = log.(x)
plot!(x_log,f_log,color=:purple,label="f_ln", framestyle=:zerolines)

plot!(-3:4, -3:4, label="(Mirror)", color=:red, linestyle=:dash, lw=1.5)
```

<a id="ln-properties"></a>
#### Natural Logarithm Properties

* **Domain:** $x \in (0, \infty)$
* **Range:** $y \in (-\infty, \infty)$
* **x-intercepts:** $(1, 0)$
* **y-intercepts:** None

[↑ Back to Table of Contents](#table-of-contents)

<a id="natural-logarithm-rules"></a>
### 3.3.2 Natural Logarithm Rules

<a id="ln-algebraic-rules"></a>
#### Natural Logarithm Algebraic Axioms

**Mathematical Concept:**
The natural logarithm obeys all standard logarithmic properties, constrained precisely to the mathematical base $e$.
$$
\begin{aligned}
\ln(e^x) &= x \quad , \quad e^{\ln(x)} = x \\
\ln(e) &= 1 \quad , \quad \ln(1) = 0 \\
\ln(xy) &= \ln(x) + \ln(y) \\
\ln\left(\frac{x}{y}\right) &= \ln(x) - \ln(y) \\
\ln(x^r) &= r \cdot \ln(x)
\end{aligned}
$$
