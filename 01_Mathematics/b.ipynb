## 1.1 Function Basics

<a id="define-a-function"></a>
### 1.1.1 Define a Function

**Mathematical Concept:**
A function maps each element in the domain to exactly one element in the range.
$$ f(x) = x^2 $$

**Modeling Objective:**
Demonstrate programmatic generation of dependent variables across an independent array for visualization.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
f(x) = x.^2
```

```julia
xV = 1:100

plot(xV,f(xV), title = "Quadratic Function Example")
```

<a id="vertical-line-test"></a>
### 1.1.2 Vertical Line Test (VLT)

**Mathematical Concept:**
The VLT verifies if a curve represents a valid function. Multiple intersections imply a non-function relation.
$$ x = c \implies y = \pm\sqrt{c} $$

**Modeling Objective:**
Geometrically demonstrate that the relation $y^2 = x$ fails the function definition via vertical line intersections.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
vline!([50], label="VLT Line at x=50", color=:red, linestyle=:dash, lw=2)
```

```julia

x_var = 0:0.1:100
y_p = sqrt.(x_var)
y_n = -sqrt.(x_var)

plot(x_var, y_p, label="y = +√x" ,color=:gray)
plot!(x_var, y_n, label="y = -√x" ,color=:gray)

vline!([25], label="VLT Line at x=25", color=:red, linestyle=:dash, lw=2)
```

<a id="standard-functions"></a>
### 1.1.3 Standard Functions

<a id="quadratic-function"></a>
#### Quadratic Function

**Mathematical Concept:**
Represents a symmetric parabola, fundamental in modeling acceleration and kinetic energy.
$$ f(x) = x^2 $$

**Modeling Objective:**
Visualize even symmetry and non-linear polynomial growth.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = x.^2

plot(x, f(x), 
     label="f(x) = x^2", 
     title="Symmetric Quadratic Curve around (0,0)", 
     color=:purple, 
     lw=2,
     framestyle=:origin,
     grid=true)
```

<a id="linear-function"></a>
#### Linear Function

**Mathematical Concept:**
Features a constant rate of change, represented geometrically by a straight line.
$$ f(x) = x $$

**Modeling Objective:**
Verify direct proportionality, forming the basis of linear system modeling.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
x = -100:0.1:100
f(x) = x

plot(x, f(x), 
     label="f(x) = x",
     color=:purple, 
     lw=2,
     framestyle=:origin,
     grid=true)
```

<a id="cubic-function"></a>
#### Cubic Function

**Mathematical Concept:**
An odd function symmetric about the origin, often used to model volumes and rapid system dynamics.
$$ f(x) = x^3 $$

**Modeling Objective:**
Illustrate point symmetry and unbounded non-linear growth in both directions.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
x = -10:0.1:10
f(x) = x.^3

plot(x, f(x), 
     label="f(x) = x^3",
     color=:purple, 
     lw=2,
     framestyle=:origin,
     grid=true)
```

<a id="square-root-function"></a>
#### Square Root Function

**Mathematical Concept:**
The inverse of the quadratic function's positive branch, restricted to a non-negative domain.
$$ f(x) = \sqrt{x} $$

**Modeling Objective:**
Demonstrate domain restriction to yield real numbers and visualize decelerating growth rates.

[⬆ Back to Table of Contents](#table-of-contents)

```julia
x = 0:0.1:10
z = sqrt.(x)

plot(x, z, 
     label="f(x) = \\sqrt{x}",
     color=:purple, 
     lw=2,
     framestyle=:origin,
     grid=true)
```
